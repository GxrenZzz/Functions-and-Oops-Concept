```python
class Audi(Car):
  def __init__(self, color, types, mileage, seat_capacity,electric, city) -> None:
    super().__init__(color,types,mileage,seat_capacity)
    self.electric = electric
    self.city  = city 

  def audi_info(self):
    print(f"Electric = {self.electric}")
    print(f"City = {self.city}")

  def full_info(self):
    self.show_basic_info()
    self.audi_info()
```

```python
c1 = Audi("Black","Petrol",22.5,4,True,"Surat")
c1.full_info()
```

```python
# Multiple Inheritance


class Father():
  def __init__(self):
    print("Father init")

  Father_name = "Piyush"

  def display_father(self):
    print(f"Father :{self.Father_name}")

class Mother():
  def __init__(self):
    print("Mother init")
  Mother_name = "Imagination"

  def display_mother(self):
    print(f"Mother :{self.Mother_name}")

class Child():
  
  def __init__(self):
    Father.__init__(self)
    print("Child init")
  Child_name = "Propa"

  def display_name(self):
    print(f"Child: {self.Child_name}")
```

```python
# Multi Level 

class Grandfather():
  def __init__(self,grandfathername):
    self.grandfathername = grandfathername

class Father(Grandfather):
  def __init__(self,Fathername ,grandfathername):
    self.Fathername = Fathername
    Grandfather.__init__(self,grandfathername)

class Son(Father):
  def __init__(self,sonname,Fathername, grandfathername):
    self.sonname = sonname 

  # Invokin the Father Class 

    Father.__init__(self,Fathername,grandfathername)

  def show_names(self):
    print(self.sonname)
    print(self.Fathername)
    print(self.grandfathername)
```

```python
s1 = Son("Piyush","Vipin","Dharam")
print(s1.grandfathername)
s1.show_names()
```
