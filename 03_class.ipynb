- Now we have a Problem that is , We're not able to get info() first cuz we need to set info , but if we forgot to do that - then the info wont apppear and raise an error 

- To solve that we use __init__. this is a constructor method which gets called instantly aftert the object is created

```python
class Students:
  def __init__(self): # constructor
      self.name = input("Enter a Name : ")  # variables are attributes
      self.age  = int(input("Enter a Age : ") )
      self.gender = input("Enter Gender : ")
      self.roll_num = int(input("Enter a Roll Number : "))

  def info(self): # method
     print(f"Name : {self.name}")
     print(f"Age : {self.age}")
     print(f"Gender: {self.gender}")
     print(f"Roll_Num : {self.roll_num}")


s1 = Students() 
```
