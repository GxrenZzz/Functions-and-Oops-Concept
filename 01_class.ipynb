### Scoping

- variable inside a function is only valid there. 
- same variable names but in diff functions work for there respective Functions

```python
def add():
  var1 = int(input("Enter num 1 = "))
  var2 = int(input("Enter num 2 = "))
  print(f"Answer : {var1 + var2}")

def sub():
    var1 = int(input("Enter num 1 = "))
    var2 = int(input("Enter num 2 = "))
    print(f"Answer : {var1 - var2}")
```

```python
add()
sub()
```
