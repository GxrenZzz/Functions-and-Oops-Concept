```python
# Encapsulation.py - Data Hiding -> Acess Modifiers - Public ,Private(__Name), Protected

# getter setter

from random import randint

class Bank():
  def __init__(self):
    self.__name = input("Enter Name : ")
    self.__acc_no = randint(10000,99999)
    self.__balance = 0 

  def display_balance(self): #getter
    print(self.__balance)

  def set_balnce(self,new_amount): #setter
    self.__balance = new_amount

  def display(self):
    print(f"NAME :{self.__name}")
    print(f"Acc : {self.__acc_no}")
    print(f"Balance : {self.__balance}")
```

```python
s1 = Bank()
# we can access members through objects
s1.display()
s1.__acc_no = 1 # this acc no shouldnt be acessible , we have to do something!!!
s1.set_balnce(10000)
s1.display_balance()
print("--------------------")
s1._Bank__balance = 1023678124679 ## Never Do This - Name Mangaling

s1.display()
```

```python
# Protected 


class Father():
  def __init__(self):
    self.father_name = input("Name : ")
    self._bank_balance = int(input("Enter Father Bank Balance : "))
    self.__phone = input("Enter Phone Model : ")

  def display_size(self):
    print(f"Father_name. = {self.father_name}")
    print(f"Bank_Balance. = {self._bank_balance}")
    print(f"Phone. = {self.__phone}")

class Child(Father):
  def __init__(self):
    super().__init__()
    self.child_name=input("Enter Child Name : ")

  def display_inf(self):
    print(F"Child name : {self.child_name}")
    print(F"Father Balance : {self._bank_balance}")
```

```python
c1 = Child()
c1.display_inf()
```

```python
# Dunder Methods 

class Father():
  def __init__(self):
    self.father_name = input("Name : ")
    self._bank_balance = int(input("Enter Father Bank Balance : "))
    self.__phone = input("Enter Phone Model : ")

  def __str__(self) -> str:
    return "I am a string dunder method"
    

  def display_size(self):
    print(f"Father_name. = {self.father_name}")
    print(f"Bank_Balance. = {self._bank_balance}")
    print(f"Phone. = {self.__phone}")
```

```python
obj = Father()
obj.display_size()
```
