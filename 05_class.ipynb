```python
class Student():
  def __init__(self):
    self.name = input("Enter Name : ")
    self.age = int(input("Enter Age : "))

  def display_info(self):
    print(f"Name : {self.name}")
    print(f"Age : {self.age}")

  def change_name(self, name:str):
    self.name = name
```

```python
s1 = Student()
s1.display_info()
s1.change_name("Cutie Poo")
s1.display_info()
```
