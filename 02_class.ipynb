### Args- For unknown Arguments (Dont know how many args in input will be passed on to Params)

```python
def add(*args):
  print(sum(args))


add(1,2,3,4)
add(1,299,121,133)
add(1,299,121,133,13342)
```

### Kwargs - For Unknown keyword Arguments 

```python
def add(**kwargs):
  print(kwargs)

add(name="Piyush",age=20,Gender="Male")
add(name="Piyush",age=20,Gender="Male",paise=10)
```

```python
# Lambda 
add_numbers = lambda n1 , n2 , n3 : n1 + n2 + n3 

print(add_numbers(10,20,30))
```

```python
# Take a argument which will be a number 
# Make a list 0 to N 

make_list = lambda n : [i for i in range (0,n+1)]

list1 = make_list(5)

print(f"{list1 = }")

```

```python
# for even return true for odd return false 


odd_even = lambda num : print("even") if num % 2 == 0 else print("odd")

odd_even(100)
```

### Anotations

```python
# not a strict annotation . just for display 

def add(x:int,y:int) -> int:
  total = x + y
  print(total)


add(10,20)
```
