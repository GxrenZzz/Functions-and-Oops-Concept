```python
# static and class method 

from datetime import datetime

class Calender():
  def __init__(self):
    self.events = [] 

  def add_Events(self,event_name):
    self.events.append(event_name)

  def display(self):
    print(f"events :{self.events}")

  @staticmethod
  def is_weekend(date:datetime):
    if date.weekday() > 4:
      print("Weekend")
    else:
      print("Weekday")


obj1 = Calender()
obj1.add_Events("Party")
obj1.add_Events("Dsa-PADHO")
obj1.add_Events("SQL")
obj1.display()


cuurent_date = datetime.now()

Calender.is_weekend(cuurent_date)
```

```python
# Class Method 

class Student:
  def __init__(self,name,age,gender) -> None:
    self.name = name
    self.age = age
    self.gender= gender

  def display(self):
    print(f"Name : {self.name}")
    print(f"AGE: {self.age}")
    print(f"Gender : {self.gender}")

  @classmethod
  def create_student_using_params(cls,name,age,gender):
    obj = cls(name,age,gender)
    return obj

  @classmethod 
  def create_student_using_file(cls,filename):
    f = open(filename ,"r")
    student_data = f.read()
    name,age,gender = student_data.split() #by unpacking
    f.close()
    obj = cls(name,age,gender)
    return obj
```

```python
# obj1 = Student("Piyush",22,"Male")
# obj1.display()

s1 = Student.create_student_using_params("Piyush",20,"Male")
s1.display()

s2 = Student.create_student_using_file("student.txt")
s2.display()
```

```python
# Decoraters -> 
from typing import *
def my_decorater(func:function) -> function:
  def wrapper() -> None:
    print("BeforeFUNC")
    func()
    print("AfterFUNC")
  return wrapper


@my_decorater
def print_hello():
  print("Hello")

print_hello()
```

```python
# generators 

def nums():
  yield 1
  yield 2
  yield 3
  yield 4
  yield 5
  yield 6

x = nums()
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```
