## Banking System

```python
from random import randint


class Bank:
  def __init__(self):
    self.account = randint(100000,999999)
    self.full_name = input("Enter Your Name")
    self.phone_number = int(input("Enter your Phone Number"))
    self.balance = 0 

  def show_info(self):
    print(f"Account Number : {self.account}")
    print(f"Full Name : {self.full_name}")
    print(f"Available Balance : {self.balance}\n")

  def show_balance(self) :
    print(f"Current Bank Balance : {self.balance}")

  def withdraw(self):
    amount = int(input("Enter Amount : "))
    if amount > self.balance:
      print("Insufficient Balance ")
    else:
      self.balance -= amount

  def deposit(self):
    amount = int(input("Enter amount : "))
    self.balance += amount


```

```python
# We use while Loop
banks = [] 

def is_acc(acc_no:int) -> bool:
  global banks
  for obj in banks:
    if obj.account == acc_no:
      return obj
  return None


while True:

  print("1.Create Account")
  print("2.Show all Bank Details")
  print("3.Deposit Amount") 
  print("4.Withdraw Amount")
  print("5.Transfer")
  print("6.Exit") 

  print("\n")

  choice = int(input("Enter a Number : "))

  if choice== 1:
    obj = Bank() 
    banks.append(obj)
    # print(banks)

  elif choice== 2:
    if len(banks) == 0:
      print("No accounts have been created yet")
    else:
      for account in banks:
        account.show_info()
        
  elif choice == 3:
    if len(banks)==0:
      print("No accounts have been created yet")
    else:
      while True:
        acc_no = int(input("Enter an Account Number to Deposit: "))

        found = False

        for obj in banks:
          if obj.account == acc_no:
            obj.deposit()
            found = True
            break

        if found:
          break

  elif choice== 4:
    if len(banks)==0:
        print("No accounts have been created yet")
    else:
      while True:
        acc_no = int(input("Enter an Account Number to Deposit: "))

        found = False

        for obj in banks:
          if obj.account == acc_no:
            obj.withdraw()
            found = True
            break

        if found == True:
          break
  elif choice== 5:
    from_acc_no = int(input("Account no from which you have to transfer"))
    to_acc_no = int(input("Account no to which you have to transfer"))
    from_acc_obj = is_acc(from_acc_no)
    to_acc_obj= is_acc(to_acc_no)
    if from_acc_obj != None and to_acc_obj != None:
      transfer_amount = int(input("Enter amount : "))
      if from_acc_obj.balance < transfer_amount:
        print("Insufficient Balance")
      else:
        from_acc_obj.balance -= transfer_amount
        to_acc_obj.balance += transfer_amount
        
    else:
      print("Invalid account Number Please Try Again")

  elif choice == 6:
    break

  else:
    print("Invalid Choice")
```
